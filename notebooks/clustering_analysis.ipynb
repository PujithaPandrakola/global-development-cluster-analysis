#Data Cleaning

```python
# Create a copy of dataset

data = df.copy()

data.head()
```

###Drop Highly Missing Columns

```python
# ==============================
# Missing Value Percentage
# ==============================

missing_percentage = (data.isnull().sum() / len(data)) * 100

missing_percentage.sort_values(ascending=False)
```

```python
# ==============================
# Drop High Missing Columns
# ==============================

threshold = 40

cols_to_drop = missing_percentage[missing_percentage > threshold].index

print("Columns to Drop:\n")
print(cols_to_drop)
```

```python
data.drop(columns=cols_to_drop, inplace=True)

print("Remaining Columns:", data.shape[1])
```

```python
# Remove constant column

data.drop(columns=['Number of Records'], inplace=True)

print("Column Removed Successfully")
```

```python
# ==============================
# Object Columns
# ==============================

object_cols = data.select_dtypes(include='object').columns

print(object_cols)
```

###Convert Numeric Object Columns

```python
# ==============================
# Convert Object Columns to Numeric
# ==============================

numeric_object_cols = [
    'GDP',
    'Health Exp/Capita',
    'Tourism Inbound',
    'Tourism Outbound'
]

for col in numeric_object_cols:

    # Remove commas
    data[col] = data[col].astype(str).str.replace(',', '')

    # Convert to numeric
    data[col] = pd.to_numeric(data[col], errors='coerce')

print("Conversion Completed")
```

```python
data.dtypes
```

```python
# ==============================
# Fill Missing Values
# ==============================

numeric_cols = data.select_dtypes(include=np.number).columns

for col in numeric_cols:
    data[col].fillna(data[col].median(), inplace=True)

print("Missing Values Handled")
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
# ==============================
# Drop Fully Null Columns
# ==============================

data.drop(columns=[
    'GDP',
    'Health Exp/Capita',
    'Tourism Inbound',
    'Tourism Outbound'
], inplace=True)

print("Fully Null Columns Removed")
```

#Data Preprocessing

```python
# ==============================
# Store Country Names
# ==============================

countries = data['Country']

countries.head()
```

```python
# ==============================
# Remove Country Column
# ==============================

X = data.drop(columns=['Country'])

X.head()
```

```python
# ==============================
# Feature Scaling
# ==============================

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
# Convert scaled data to DataFrame

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

X_scaled.head()
```

```python
# Mean should be close to 0
X_scaled.mean().round(2)
```

```python
# Std should be close to 1
X_scaled.std().round(2)
```

#KMeans Clustering Model Building

```python
# ==============================
# Import Clustering Libraries
# ==============================

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
# ==============================
# Elbow Method
# ==============================

wcss = []

K = range(2, 11)

for k in K:

    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )

    kmeans.fit(X_scaled)

    wcss.append(kmeans.inertia_)
```

```python
# ==============================
# Plot Elbow Curve
# ==============================

plt.figure(figsize=(8,5))

plt.plot(K, wcss, marker='o')

plt.title('Elbow Method')

plt.xlabel('Number of Clusters')

plt.ylabel('WCSS')

plt.show()
```

```python
# ==============================
# Silhouette Scores
# ==============================

silhouette_scores = []

for k in range(2, 11):

    kmeans = KMeans(
        n_clusters=k,
        random_state=42
    )

    cluster_labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        cluster_labels
    )

    silhouette_scores.append(score)
```

```python
# ==============================
# Plot Silhouette Scores
# ==============================

plt.figure(figsize=(8,5))

plt.plot(
    range(2,11),
    silhouette_scores,
    marker='o'
)

plt.title('Silhouette Scores')

plt.xlabel('Number of Clusters')

plt.ylabel('Silhouette Score')

plt.show()
```

```python
# Display scores

for k, score in zip(range(2,11), silhouette_scores):

    print(f"K = {k}  -->  Silhouette Score = {score:.4f}")
```

```python
# ==============================
# Final KMeans Model
# ==============================

kmeans = KMeans(
    n_clusters=5,
    random_state=42
)

clusters = kmeans.fit_predict(X_scaled)
```

```python
# Add cluster labels

data['Cluster'] = clusters

data.head()
```

```python
# Cluster counts

data['Cluster'].value_counts()
```

```python
# ==============================
# Import PCA
# ==============================

from sklearn.decomposition import PCA
```

```python
# ==============================
# PCA Transformation
# ==============================

pca = PCA(n_components=2)

pca_data = pca.fit_transform(X_scaled)
```

```python
# PCA DataFrame

pca_df = pd.DataFrame(
    data=pca_data,
    columns=['PC1', 'PC2']
)

pca_df['Cluster'] = clusters

pca_df.head()
```

```python
# ==============================
# Cluster Visualization
# ==============================

plt.figure(figsize=(10,6))

sns.scatterplot(
    x='PC1',
    y='PC2',
    hue='Cluster',
    palette='Set1',
    data=pca_df,
    s=100
)

plt.title('KMeans Clustering Visualization')

plt.show()
```

```python
# ==============================
# Cluster Analysis
# ==============================

cluster_analysis = data.groupby('Cluster').mean(numeric_only=True)

cluster_analysis
```

#Hierarchical Clustering


```python
# ==============================
# Import Libraries
# ==============================

from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
# ==============================
# Dendrogram
# ==============================

plt.figure(figsize=(15,6))

linked = linkage(
    X_scaled,
    method='ward'
)

dendrogram(linked)

plt.title('Dendrogram')

plt.xlabel('Data Points')

plt.ylabel('Distance')

plt.show()
```

```python
# ==============================
# Hierarchical Clustering
# ==============================

hc = AgglomerativeClustering(
    n_clusters=5
)

hc_clusters = hc.fit_predict(X_scaled)
```

```python
# Add labels

data['HC_Cluster'] = hc_clusters

data.head()
```

```python
# Cluster distribution

data['HC_Cluster'].value_counts()
```

```python
# ==============================
# Hierarchical Silhouette Score
# ==============================

hc_score = silhouette_score(
    X_scaled,
    hc_clusters
)

print("Hierarchical Clustering Silhouette Score:", hc_score)
```

```python
# KMeans score

kmeans_score = silhouette_score(
    X_scaled,
    clusters
)

print("KMeans Silhouette Score:", kmeans_score)
```

```python
# ==============================
# Comparative Analysis
# ==============================

comparison = pd.DataFrame({

    'Model': [
        'KMeans',
        'Hierarchical Clustering'
    ],

    'Silhouette Score': [
        kmeans_score,
        hc_score
    ]
})

comparison
```
